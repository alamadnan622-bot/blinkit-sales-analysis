## **Fill missing Item Weight values with median**

```python
df['Item Weight'] = df['Item Weight'].fillna(df['Item Weight'] . median())
```

```python
# Check missing values again
df.isnull().sum()
```

```python
print("Total missing values:", df.isnull().sum().sum())
```

**Data Cleaning**

```python
print(df['Item Fat Content'].unique())
```

```python
df['Item Fat Content'] = df['Item Fat Content'].str.strip()

df['Item Fat Content'] = df["Item Fat Content"].replace({
    'low fat': 'Low Fat',
    'LF': 'Low Fat',
    'reg': 'Regular'
})
```

```python
print(df['Item Fat Content'].unique())
```

# **KPI's Requirment**

```python
# Total Sales
total_sales = df['Sales'].sum()

# Avrage Sales
avrage_sales = df['Sales'].mean()

# Number Of Items
no_of_items = df['Sales'].count()

# Avrage rating
avg_rating = df['Rating'].mean()

#Display
print("Total Sales:", total_sales)
print("Avrage Sales:", avrage_sales)
print("Number Of Item :",no_of_items)
print("Avrage rating:", avg_rating)
```

# **Chart’s Requirements**

**1.Total Sales by Fat Content**

```python

# Total Sales by Fat Content
fat_sales = df.groupby('Item Fat Content')['Sales'].sum()

print(fat_sales)
```

```python
# Donut chart
plt.figure(figsize=(7, 7))

fat_sales.plot(
    kind='pie',
    autopct='%1.0f%%',
    startangle=90,
    wedgeprops={'width': 0.4}
)

plt.title('Total Sales by Fat Content')
plt.ylabel('')
plt.show()
```

# **3. Fat Content by Outlet for Total Sales**

```python

# Group sales by Outlet and Fat Content
sales_by_outlet_fat = df.pivot_table(
    index='Outlet Identifier',
    columns='Item Fat Content',
    values='Sales',
    aggfunc='sum',
    fill_value=0
)

print(sales_by_outlet_fat)
```

```python

sales_by_outlet_fat.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 6)
)

plt.title('Fat Content by Outlet for Total Sales')
plt.xlabel('Outlet Identifier')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.legend(title='Fat Content')
plt.tight_layout()
plt.show()
```

```python

fat_kpis = df.groupby('Item Fat Content').agg(
    Average_Sales=('Sales', 'mean'),
    Number_of_Items=('Item Identifier', 'nunique'),
    Average_Rating=('Rating', 'mean'),
    Total_Sales=('Sales', 'sum')
)

display(fat_kpis.round(2))
```

# **4. Total Sales by Outlet Establishment**

```python

# Total Sales by Outlet Establishment Year
sales_by_year = df.groupby(
    'Outlet Establishment Year'
)['Sales'].sum().sort_index()

print(sales_by_year)
```

```python

plt.figure(figsize=(10, 5))

plt.plot(
    sales_by_year.index,
    sales_by_year.values,
    marker='o'
)

plt.title('Total Sales by Outlet Establishment Year')
plt.xlabel('Outlet Establishment Year')
plt.ylabel('Total Sales')
plt.xticks(sales_by_year.index, rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

# **5. Sales by Outlet Size:**

```python

# Total Sales by Outlet Size
sales_by_size = df.groupby('Outlet Size')['Sales'].sum()

print(sales_by_size.sort_values(ascending=False))
```

```python

plt.figure(figsize=(7, 7))

plt.pie(
    sales_by_size,
    labels=sales_by_size.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'width': 0.4}
)

plt.title('Total Sales by Outlet Size')
plt.show()
```

# **6. Sales by Outlet Location**

```python

# Sales by Outlet Location
sales_by_location = (
    df.groupby('Outlet Location Type')['Sales']
    .sum()
    .sort_values(ascending=False)
)

print(sales_by_location)
```

```python

plt.figure(figsize=(8, 5))

sales_by_location.sort_values().plot(
    kind='barh',
    color='orange'
)

plt.title('Sales by Outlet Location')
plt.xlabel('Total Sales')
plt.ylabel('Outlet Location Type')
plt.tight_layout()
plt.show()
```
